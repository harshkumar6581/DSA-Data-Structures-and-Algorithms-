```python
# 11. Write a program to check whether the triangle is equilateral, scalene, or isosceles
a = int(input('Enter a first number '))
b = int(input('Enter a first number '))
c = int(input('Enter a first number '))

if a == b and b == c and c==a:
    print(a,b,c,'Equilateral Triangle ')
elif a == b or b == c or c==a: 
    print(a,b,c,'Isosceles Triangle')
else:
    print(a,b,c,'Scalene Triangle')
```
