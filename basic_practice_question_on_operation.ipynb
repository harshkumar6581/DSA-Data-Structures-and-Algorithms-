```python
# 11. Write a program to calculate sum of first 'N' even natural numbers
n = int(input('enter N :'))
 
sum = 0
for i in range(1, n + 1):
    sum = sum +(2 * i)
print("Sum of the N even natural numbers is",sum)
```

```python
## 12. Write a program to calculate sum of first 'N' odd natural numbers
n = int(input('Enter natural no '))
sum = 0
for i in range (1,n+1):
    sum = sum + (2 * i -1)
print(sum)
```

```python
# 15. Write a program to print 'XOR' of 'N' numbers
n = int(input("Enter N: "))

result = 0

for i in range(n):
    num = int(input("Enter number: "))
    result = result ^ num

print("XOR =", result)
```
